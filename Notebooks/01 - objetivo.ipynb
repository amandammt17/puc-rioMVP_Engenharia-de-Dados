## 1. Contexto e Definição do Problema (O "Porquê")
No ambiente altamente competitivo do comércio eletrônico, a pontualidade na entrega é um dos pilares centrais da retenção de clientes. Entregas atrasadas não apenas geram insatisfação imediata, mas também sobrecarregam os canais de atendimento e impactam negativamente a reputação da marca. 

O problema central que este projeto visa resolver é: **Identificar os principais estrangulamentos na cadeia de distribuição que causam atrasos nas entregas e compreender como essas falhas logísticas impactam diretamente a métrica de satisfação do cliente.** 

Sem uma visão clara de onde o tempo está sendo perdido — se na aprovação, na separação pelo vendedor ou no transporte —, a empresa atua às cegas, tentando resolver problemas de SLA (Acordo de Nível de Serviço) sem atacar a causa raiz.

## 2. Objetivo do MVP
O objetivo deste MVP de Engenharia de Dados é construir um pipeline (ETL/ELT) que extraia dados transacionais brutos e os transforme em um modelo de dados analítico (como um *Star Schema*). Esse pipeline será desenhado especificamente para fornecer uma base de dados limpa, confiável e otimizada que permita às equipes de Inteligência de Negócios (BI) e Logística responderem de forma rápida e precisa às perguntas de negócio definidas, conectando os fluxos físicos de mercadorias com o sentimento do consumidor.

## 3. Perguntas de Negócio
Para garantir que a coleta, transformação e modelagem dos dados tenham um propósito claro, o pipeline será construído para responder às seguintes perguntas fundamentais:

1. **Visão Geográfica:** Qual é a taxa média de atraso (diferença entre data estimada e data real de entrega) agrupada por Estado (UF) de origem e destino? Existem rotas críticas evidentes?
2. **Localização do Gargalo:** Onde ocorre a maior perda de tempo logístico? É no processamento interno (tempo entre a aprovação do pagamento e a postagem na transportadora) ou no trânsito (tempo entre a postagem e a entrega ao cliente)?
3. **Impacto na Satisfação:** Qual é a correlação direta entre o volume de dias de atraso e a nota de avaliação do pedido? Quantas das avaliações de 1 e 2 estrelas estão associadas a quebras no prazo de entrega?
4. **Performance de Parceiros:** Existem vendedores (*sellers*) específicos que falham sistematicamente no prazo de expedição, prejudicando o fluxo logístico downstream (etapas seguintes)?

*(Nota arquitetural: Estas perguntas justificam a necessidade de calcularmos deltas de tempo (dias/horas) entre os diversos eventos de um pedido durante a fase de transformação dos dados).*

## 4. Descrição da Base de Dados
Para responder a essas perguntas, o projeto utilizará o **Brazilian E-Commerce Public Dataset by Olist** (disponível no Kaggle). Trata-se de um conjunto de dados real e anonimizado do e-commerce brasileiro, contendo informações de cerca de 100 mil pedidos realizados entre 2016 e 2018.

A base é altamente relacional e simula um ambiente corporativo complexo, dividida em vários arquivos CSV. Para este MVP focado em logística, utilizaremos as seguintes tabelas principais:

*   **`olist_orders_dataset` (A tabela central):** Essencial para o projeto, pois contém todos os *timestamps* vitais do fluxo do pedido: data da compra, data de aprovação, data de postagem (envio), data de entrega real e data estimada de entrega.
*   **`olist_order_reviews_dataset`:** Contém as notas de satisfação (1 a 5 estrelas) dadas pelos clientes após a experiência de compra, além das datas em que as avaliações foram criadas e respondidas.
*   **`olist_order_items_dataset`:** Permite conectar os pedidos aos vendedores (`seller_id`), fundamental para mapear os responsáveis por gargalos na expedição.
*   **`olist_customers_dataset` e `olist_sellers_dataset`:** Fornecem os dados demográficos (CEP, Cidade, Estado) de quem compra e de quem vende, viabilizando as análises de rotas críticas e problemas geográficos de entrega.

O desafio da Engenharia de Dados será unificar essas tabelas, tratar dados nulos ou inconsistentes (como pedidos sem data de entrega registrada) e criar *features* analíticas temporais que não existem na base bruta.
